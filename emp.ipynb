```python
#Create a Pandas DataFrame and display the *top 3 highest-paid employees*.
import pandas as pd

data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Emp_Name": ["Harsha", "Anjali", "Rahul", "Sneha", "Arjun", "Priya", "Kiran", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Marketing", "Finance", "HR", "IT"],
    "Salary": [65000, 48000, 72000, 58000, 81000, 69000, 55000, 76000]
}

df=pd.DataFrame(data)
print(df)
```

```python
print(df.nlargest(3,'Salary'))
```

```python
print(df.nsmallest(5,'Salary'))
```

```python
#Group employees by *department* and display the *total salary*
df.groupby('Department')['Salary'].sum()
```
