```python
#Create a NumPy array and find the sum, mean, maximum, and minimum values.
import numpy as np
arr=np.array([2,4,6,8,10])
print(f"The sum of the array: {arr.sum()}")
print(f"The mean of the array: {arr.mean()}")
print(f"The maximam number of the array {arr.max()}")
print(f"The minimum of the array: {arr.min()}")
```
